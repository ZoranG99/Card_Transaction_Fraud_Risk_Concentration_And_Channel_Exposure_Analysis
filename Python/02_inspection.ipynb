# Data Inspection

## 1. Import Libraries and Load Data

```python
import pandas as pd
import glob
import os
from pathlib import Path

pd.set_option('display.max_columns', None)  # show all columns

# Folder with the raw CSVs
ROOT = Path.cwd().parent                       
raw_path = ROOT / "Dataset" / "Raw"

# Load every CSV into a dict: {table_name: dataframe}
tables = {}
for file in glob.glob(os.path.join(raw_path, "*.csv")):
    name = os.path.basename(file).replace(".csv", "")
    tables[name] = pd.read_csv(file)

print(f"Loaded {len(tables)} tables:")
for name, df in tables.items():
    print(f"  {name}: {df.shape[0]} rows, {df.shape[1]} cols")
```

## 2. Preview Table Snapshots

```python
for name, df in tables.items():
    print(f"\n{'='*70}\n{name}\n{'='*70}")
    display(df.head())
```

## 3. Inspect Data Types and Structure

```python
for name, df in tables.items():
    print(f"\n{'='*70}\n{name}\n{'='*70}")
    df.info()
```

## 4. Audit Missing Values

```python
for name, df in tables.items():
    print(f"\n{'='*70}\n{name} — nulls per column\n{'='*70}")
    print(df.isnull().sum())
```

## 5. Detect Duplicate Records

```python
for name, df in tables.items():
    dupes = df.duplicated().sum()
    print(f"{name}: {dupes} duplicate rows")
```

## 6. Explore Numeric Summary Statistics

```python
for name, df in tables.items():
    print(f"\n{'='*70}\n{name}\n{'='*70}")
    display(df.describe())
```

## 7. Explore Categorical Summary Statistics

```python
for name, df in tables.items():
    print(f"\n{'='*70}\n{name}\n{'='*70}")
    display(df.describe(include=['object', 'category', 'string']))
```

## 8. Check Column Cardinality

```python
for name, df in tables.items():
    print(f"\n{'='*70}\n{name} (total rows: {len(df)})\n{'='*70}")
    nunique = df.nunique()    
    print(nunique)
```

## 9. Review Common Value Patterns

```python
for name, df in tables.items():    
    print(f"\n{'='*70}\n{name} \n{'='*70}")
    for col in df.columns:        
        print(df[col].value_counts().head(10))
        print()
```

## 10. Check Date Range (transactions_data)

```python
transactions = tables["transactions_data"]  
transactions["date"] = pd.to_datetime(transactions["date"])

print("Min date:", transactions["date"].min())
print("Max date:", transactions["date"].max())
print("Total span:", transactions["date"].max() - transactions["date"].min())

print("\nRows per year:")
print(transactions["date"].dt.year.value_counts().sort_index())
```
