# Json Conversion

```python
import json
import pandas as pd
from pathlib import Path
from IPython.display import display

RAW_DIR = Path("../Dataset/Raw")

# --- MCC codes ---
with open(RAW_DIR / "mcc_codes.json") as f:
    mcc = json.load(f)
mcc_df = pd.DataFrame(list(mcc.items()), columns=["mcc_code", "description"])
mcc_df.to_csv(RAW_DIR / "mcc_codes.csv", index=False)

# --- Fraud labels ---
with open(RAW_DIR / "train_fraud_labels.json") as f:
    fraud = json.load(f)

fraud_dict = fraud["target"]  
fraud_df = pd.DataFrame(list(fraud_dict.items()), columns=["transaction_id", "is_fraud"])
fraud_df.to_csv(RAW_DIR / "fraud_labels.csv", index=False)

print(f"mcc_df shape: {mcc_df.shape}")
display(mcc_df.head())

print(f"fraud_df shape: {fraud_df.shape}")
display(fraud_df.head())
```
